```python
%sql
SELECT
    f.transaction_id,
    c.customer_name,
    c.city,
    c.state,
    p.product_name,
    p.category,
    p.product_segment,
    cal.full_date,
    cal.month_name,
    cal.quarter,
    cal.year,
    cal.is_weekend,
    f.quantity,
    f.selling_price,
    f.discount_amount,
    f.gross_amount,
    f.net_sales_amount,
    f.payment_mode,
    f.sales_channel
FROM retail_q.retail_gold.fact_sales f
LEFT JOIN retail_q.retail_gold.dim_customer c
    ON f.customer_id = c.customer_id
LEFT JOIN retail_q.retail_gold.dim_product p
    ON f.product_id = p.product_id
LEFT JOIN retail_q.retail_gold.dim_calendar cal
    ON f.date_key = cal.date_key
LIMIT 20
```
