```python
%sql

CREATE SCHEMA IF NOT EXISTS retail_q.retail_semantic;

CREATE OR REPLACE VIEW retail_q.retail_semantic.retail_metrics
WITH METRICS
LANGUAGE YAML
AS $$
version: 1.1

source: retail_q.retail_gold.fact_sales
comment: Retail metrics for analyzing sales transactions, revenue, and product performance

joins:
  - name: product
    source: retail_q.retail_gold.dim_product
    on: source.product_id = product.product_id
  - name: calendar
    source: retail_q.retail_gold.dim_calendar
    on: source.date_key = calendar.date_key
  - name: customer
    source: retail_q.retail_gold.dim_customer
    on: source.customer_id = customer.customer_id

dimensions:
  - name: Transaction Date
    expr: calendar.full_date
    display_name: Transaction Date
    format:
      type: date
      date_format: year_month_day
    synonyms:
      - date
      - sale date
      - transaction day

  - name: Year
    expr: calendar.year
    display_name: Year

  - name: Quarter
    expr: calendar.quarter
    display_name: Quarter

  - name: Month Name
    expr: calendar.month_name
    display_name: Month
    synonyms:
      - month

  - name: Is Weekend
    expr: calendar.is_weekend
    display_name: Is Weekend

  - name: Product Category
    expr: product.category
    display_name: Product Category
    synonyms:
      - category

  - name: Product Subcategory
    expr: product.subcategory
    display_name: Product Subcategory
    synonyms:
      - subcategory

  - name: Product Segment
    expr: product.product_segment
    display_name: Product Segment
    synonyms:
      - segment

  - name: Product Brand
    expr: product.brand
    display_name: Brand
    synonyms:
      - brand

  - name: Payment Mode
    expr: source.payment_mode
    display_name: Payment Mode
    synonyms:
      - payment method
      - payment type

  - name: Sales Channel
    expr: source.sales_channel
    display_name: Sales Channel
    synonyms:
      - channel

  - name: Customer Type
    expr: customer.customer_type
    display_name: Customer Type
    synonyms:
      - customer segment

  - name: Customer Name
    expr: customer.customer_name
    display_name: Customer Name
    synonyms:
      - customer

  - name: City
    expr: customer.city
    display_name: City
    synonyms:
      - customer city

  - name: State
    expr: customer.state
    display_name: State
    synonyms:
      - customer state

  - name: Country
    expr: customer.country
    display_name: Country
    synonyms:
      - customer country

  - name: Industry
    expr: customer.industry
    display_name: Industry
    synonyms:
      - customer industry

measures:
  - name: Transaction Count
    expr: COUNT(1)
    display_name: Transaction Count
    format:
      type: number
      decimal_places:
        type: exact
        places: 0
    synonyms:
      - transactions
      - count
      - order count

  - name: Total Revenue
    expr: SUM(net_sales_amount)
    display_name: Total Revenue
    format:
      type: currency
      currency_code: INR
      decimal_places:
        type: exact
        places: 2
    synonyms:
      - revenue
      - sales
      - total sales
      - net sales

  - name: Total Gross Amount
    expr: SUM(gross_amount)
    display_name: Total Gross Amount
    format:
      type: currency
      currency_code: INR
      decimal_places:
        type: exact
        places: 2
    synonyms:
      - gross revenue
      - gross sales

  - name: Total Quantity Sold
    expr: SUM(quantity)
    display_name: Total Quantity
    format:
      type: number
      decimal_places:
        type: exact
        places: 0
    synonyms:
      - quantity
      - units sold

  - name: Total Discount
    expr: SUM(discount_amount)
    display_name: Total Discount
    format:
      type: currency
      currency_code: INR
      decimal_places:
        type: exact
        places: 2
    synonyms:
      - discounts

  - name: Average Transaction Value
    expr: SUM(net_sales_amount) / COUNT(1)
    display_name: Avg Transaction Value
    format:
      type: currency
      currency_code: INR
      decimal_places:
        type: exact
        places: 2
    synonyms:
      - avg revenue
      - average order value
      - aov

  - name: Unique Customers
    expr: COUNT(DISTINCT customer_id)
    display_name: Unique Customers
    format:
      type: number
      decimal_places:
        type: exact
        places: 0
    synonyms:
      - customer count
      - distinct customers

  - name: Unique Products
    expr: COUNT(DISTINCT product_id)
    display_name: Unique Products
    format:
      type: number
      decimal_places:
        type: exact
        places: 0
    synonyms:
      - product count
$$
```
